## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The outcome used to evaluate the ranking is observed content decline. A page is considered declining when the dataset records its observed trend direction as "down".

This is an observed outcome from the data rather than a priority rule created by the model. The ranking will use available content and performance signals to prioritize pages, and the observed decline outcome will be used to check whether high-priority pages are actually more likely to be declining.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Target/proxy column: trend_direction")
print("\nTrend direction values:")
print(df["trend_direction"].value_counts(dropna=False))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The main success metric will be Precision@50.

The team can only review a limited number of pages first, so the ranking should place as many actually declining pages as possible near the top. Precision@50 measures the proportion of the top 50 recommended pages that are actually observed as declining.

A higher Precision@50 means the priority list is more useful for the review team.

```python
print("Primary success metric: Precision@50")
print("Meaning: proportion of the top 50 ranked pages that are actually declining.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could use one signal, such as a large decrease in impressions, to create a refresh list. However, content performance depends on multiple signals, including impressions, clicks, position, query behavior, and other historical patterns.

ML can combine these signals and learn patterns from observed outcomes instead of relying on one manually chosen threshold. The model is therefore intended to support and prioritize human review rather than automatically decide which pages should be refreshed.

```python
features_to_check = [
    "imp_last30",
    "imp_prev30",
    "clk_last30",
    "pos_last30"
]

available_features = [col for col in features_to_check if col in df.columns]

print("Signals available for the ML problem:")
print(available_features)
```
