## My Lane: Content Refresh Prioritization

I chose the Content Refresh Prioritization lane because websites often contain thousands of pages, but only a small portion need immediate attention. ## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

Which content pages are most likely to experience declining search performance and should be prioritized for refresh?

### Decision

The model helps decide which content pages should be reviewed first.

### Action

The SEO or content team reviews the recommended pages and updates the content by improving quality, adding new information, fixing outdated sections, or optimizing SEO elements.

### Cost of a Wrong Decision

If the model recommends a healthy page (false positive), the team may spend time updating content that did not need attention.

If the model misses a truly declining page (false negative), that page may continue losing search visibility and traffic before anyone notices.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining = (df["trend_direction"].str.lower() == "down").sum()
declining_rate = (df["trend_direction"].str.lower() == "down").mean()

print("Total Pages:", len(df))
print("Declining Pages:", declining)
print("Declining Rate:", round(declining_rate,3))
```

### Initial observations

The starter dataset contains **30,000** content pages.

Approximately **54.2%** of the pages are labeled as declining.

In the full warehouse notebook, more than **111,000** content items had enough history to build features for modeling. These numbers suggest that content decline is common enough to justify building a prioritization system instead of reviewing every page manually.
